## TODO 2:

Translate these English sentences into First-Order Logic


1.   Every student who studies hard will pass every exam.
2.   There exists a teacher who teaches every subject to some student.
3.   Every person who loves someone will eventually be loved by someone.
4.   No country that has a nuclear weapon is trusted by all other countries.
5.   Every employee who is managed by someone who is also an employee gets a bonus if the company makes a profit.
6.   No scientist publishes a paper unless it has been reviewed by at least two other scientists.





Predicates have their ordinary meanings. Variables range over all objects, so unary predicates restrict their types. In $Teaches(t,s,x)$, teacher $t$ teaches subject $s$ to student $x$; $Trusts(d,c)$ means country $d$ trusts country $c$; $Manages(m,e)$ means $m$ manages $e$; $Reviewed(r,p)$ means $r$ has reviewed paper $p$.

1. **Every student who studies hard will pass every exam.**

   $$\forall x\;\left[(Student(x)\land StudiesHard(x))\Rightarrow\forall e\;(Exam(e)\Rightarrow Passes(x,e))\right].$$

2. **There exists a teacher who teaches every subject to some student.**

   $$\exists t\;\left[Teacher(t)\land\forall s\;\left(Subject(s)\Rightarrow\exists x\;(Student(x)\land Teaches(t,s,x))\right)\right].$$

   The same teacher teaches all subjects; the student may depend on the subject. If "some student" means one fixed student for all subjects, the stronger reading is $\exists t\exists x\;[Teacher(t)\land Student(x)\land\forall s\;(Subject(s)\Rightarrow Teaches(t,s,x))]$.

3. **Every person who loves someone will eventually be loved by someone.**

   Using the usual introductory, time-abstracted predicates:

   $$\forall x\;\left[(Person(x)\land\exists y\;(Person(y)\land Loves(x,y)))\Rightarrow\exists z\;(Person(z)\land Loves(z,x))\right].$$

   To express **eventually** explicitly in First-Order Logic, introduce time: $LovesAt(x,y,t)$ means $x$ loves $y$ at time $t$, and $u>t$ means $u$ is later than $t$. Then the time-sensitive translation is:

   $$\forall x\forall t\;\left[(Person(x)\land Time(t)\land\exists y\;(Person(y)\land LovesAt(x,y,t)))\Rightarrow\exists z\exists u\;(Person(z)\land Time(u)\land u>t\land LovesAt(z,x,u))\right].$$

   The future person $z$ need not be the person $y$ whom $x$ loves. The binary-predicate version above omits the timing condition.

4. **No country that has a nuclear weapon is trusted by all other countries.**

   $$\forall c\;\left[(Country(c)\land\exists w\;(NuclearWeapon(w)\land Has(c,w)))\Rightarrow\neg\forall d\;((Country(d)\land d\ne c)\Rightarrow Trusts(d,c))\right].$$

   Equivalently, every country with a nuclear weapon has at least one **other** country that does not trust it:

   $$\forall c\;\left[(Country(c)\land\exists w\;(NuclearWeapon(w)\land Has(c,w)))\Rightarrow\exists d\;(Country(d)\land d\ne c\land\neg Trusts(d,c))\right].$$

5. **Every employee who is managed by someone who is also an employee gets a bonus if the company makes a profit.**

   Let constant $c$ denote the company under discussion, with $Company(c)$.

   $$\forall e\;\left[(Employee(e)\land\exists m\;(Employee(m)\land Manages(m,e))\land MakesProfit(c))\Rightarrow GetsBonus(e)\right].$$

   Profit is a sufficient condition for the bonus; the sentence does not say that a bonus is possible only when there is a profit.

6. **No scientist publishes a paper unless it has been reviewed by at least two other scientists.**

   $$\forall s\forall p\;\left[(Scientist(s)\land Paper(p)\land Publishes(s,p))\Rightarrow\exists r_1\exists r_2\;\left(Scientist(r_1)\land Scientist(r_2)\land r_1\ne r_2\land r_1\ne s\land r_2\ne s\land Reviewed(r_1,p)\land Reviewed(r_2,p)\right)\right].$$

   The two reviewers are distinct from each other and from the publishing scientist. Review is necessary for publication, not sufficient to guarantee publication.
